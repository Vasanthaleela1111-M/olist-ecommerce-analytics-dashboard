This notebook performs the raw schema discovery, key cardinalities audit, missing value profiling, and relationship topology verification across all **9 raw CSV datasets**.

```python
import os
import sys
import pandas as pd
import numpy as np

sys.path.append('..')
from src.load_data import load_all_datasets, validate_schemas

# Load all 9 datasets
raw_datasets = load_all_datasets('../data')
validate_schemas(raw_datasets)
```

### Schema Audit Summary Table

```python
audit_rows = []
for name, df in raw_datasets.items():
    audit_rows.append({
        'dataset': name,
        'row_count': len(df),
        'column_count': len(df.columns),
        'null_count': int(df.isnull().sum().sum()),
        'duplicate_rows': int(df.duplicated().sum())
    })
display(pd.DataFrame(audit_rows))
```
